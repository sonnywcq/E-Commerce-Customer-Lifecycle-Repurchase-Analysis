```python
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', None)
```

```python
DATA_PATH="/Users/wuchaoqun/Desktop/Project1/raw data/"
```

```python
customers = pd.read_csv(
    DATA_PATH+"olist_customers_dataset.csv"
)


orders = pd.read_csv(
    DATA_PATH+"olist_orders_dataset.csv"
)


items = pd.read_csv(
    DATA_PATH+"olist_order_items_dataset.csv"
)


payments = pd.read_csv(
    DATA_PATH+"olist_order_payments_dataset.csv"
)


products = pd.read_csv(
    DATA_PATH+"olist_products_dataset.csv"
)
```

```python
datasets={
    "customers":customers,
    "orders":orders,
    "items":items,
    "payments":payments,
    "products":products
}


for name,df in datasets.items():

    print(
        name,
        df.shape
    )
```

```python
for name,df in datasets.items():

    print("\n",name)

    print(
        df.isnull().sum()
    )
```

```python
orders['order_purchase_timestamp']=pd.to_datetime(
    orders['order_purchase_timestamp']
)


orders['order_delivered_customer_date']=pd.to_datetime(
    orders['order_delivered_customer_date']
)
```

```python
orders_clean=orders[
    orders['order_status']=="delivered"
].copy()
```

```python
customer_orders = orders_clean.merge(

    customers,

    on="customer_id",

    how="left"

)


customer_orders.head()
```

```python
order_detail = customer_orders.merge(

    items,

    on="order_id",

    how="left"

)
```

```python
final_data = order_detail.merge(

    payments,

    on="order_id",

    how="left"

)
```

```python
final_data.head()
```

```python
final_data['order_amount']=(
    final_data['price']
    +
    final_data['freight_value']
)
```

```python
analysis_data = final_data[

[
'order_id',

'customer_unique_id',

'order_purchase_timestamp',

'product_id',

'price',

'freight_value',

'order_amount',

'payment_value',

'customer_city',

'customer_state'

]

]
```

```python
analysis_data.head()
```

```python
analysis_data.to_csv(

"/Users/wuchaoqun/Desktop/Project1/raw data/customer_order_analysis.csv",

index=False

)
```
