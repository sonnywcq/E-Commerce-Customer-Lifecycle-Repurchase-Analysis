```python
import pandas as pd

import numpy as np
```

```python
df=pd.read_csv(

"/Users/wuchaoqun/Desktop/Project1/raw data/customer_order_analysis.csv"

)


df['order_purchase_timestamp']=pd.to_datetime(

df['order_purchase_timestamp']

)
```

```python
df['order_month']=(

df['order_purchase_timestamp']

.dt

.to_period('M')

)
```

```python
df['first_month']=(

df.groupby(

'customer_unique_id'

)['order_month']

.transform('min')

)
```

```python
df['cohort_index']= (

df['order_month']

-

df['first_month']

).apply(

lambda x:x.n

)
```

```python
cohort_data=df.groupby(

[

'first_month',

'cohort_index'

]

)[

'customer_unique_id'

].nunique().reset_index()
```

```python
cohort_table = cohort_data.pivot(

index='first_month',

columns='cohort_index',

values='customer_unique_id'

)
```

```python
cohort_table.head()
```

```python
retention = cohort_table.divide(

cohort_table.iloc[:,0],

axis=0

)
```

```python
retention.round(3)*100
```

```python
retention.to_csv(

"/Users/wuchaoqun/Desktop/Project1/raw data/cohort_retention.csv"

)
```
